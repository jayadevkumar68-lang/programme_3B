```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load dataset
df = pd.read_csv("hospital.csv")

# Display dataset
print(df.head())

# Set plot style
sns.set_style("whitegrid")

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Line plot
sns.lineplot(
    data=df,
    x="Month",
    y="Emergency Cases",
    marker="o",
    color="blue",
    ax=axes[0]
)

axes[0].set_title("Monthly Emergency Cases Trend")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Emergency Cases")

# Bar plot
sns.barplot(
    data=df,
    x="Month",
    y="Emergency Cases",
    color="green",
    ax=axes[1]
)

axes[1].set_title("Emergency Cases by Month")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Emergency Cases")

plt.tight_layout()
plt.show()
```
